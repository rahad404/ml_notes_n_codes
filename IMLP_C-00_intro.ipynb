```python
eye = np.eye(4)
print(f"eye:\n{eye}")
```

### Sparse matrix
sparse matrix are used whenever we want to store a 2D array that contains mostly zero

_convert numpy array to a scipy sparse matrix in csr format_

```python
sparse_matrix = sparse.csr_matrix(eye)
print(f"sparse_matrix:\n{sparse_matrix}")
```

```python
arr = np.array([[1,0,0],[0,1,0],[1,1,3]])
sp_arr = sparse.csr_matrix(arr)
print(f"new sp array:\n{sp_arr}")
```

Usually it's not possible to create dense representation of sparse data{wont fit into memory}, 
we create sparse representation directly 

_sparse matrix using the  __COO__ format_

```python
data = np.ones(4)
row_indices = np.arange(4)
col_indeces = np.arange(4)
eye_coo = sparse.coo_matrix((data, (row_indices, col_indeces)))
print(f"COO representation:\n{eye_coo}")
```
