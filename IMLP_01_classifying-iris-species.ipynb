### A First Application: Classifying Iris Species

have measurement of known correct species of iris
so, this is a __Supervised Learning__ problem
this is also a __classification__ problem

* the possible outputs are called _classes_
* every iris(in the dataset) belong to one of three classes
* so, this problem is a three-classes classification problem.
* for particular datapoint the species belong to is called its _label_


```python
from sklearn.datasets import load_iris
iris_dataset = load_iris()
```

```python
print(f"Key of dataset: \n{iris_dataset.keys()}")
print(iris_dataset["DESCR"][:193]+"\n...")
```

```python
print(f"Target names:\n{iris_dataset["target_names"]}")
print(f"\nFeature names:\n{iris_dataset["feature_names"]}")
print(f"\nType of data: \n{type(iris_dataset["data"])}")
print(f"\nShape of data: \n{iris_dataset["data"].shape}")
print(f"\nFirst five column of data:\n{iris_dataset["data"][:5]}")
```

The meanings of the numbers are given by the iris['target_names'] array:
* 0 means setosa
* 1 means versicolor
* 2 means virginica

```python
print(f"Type of target: {type(iris_dataset['target'])}")
print(f"\nShape of target: {iris_dataset['target'].shape}")
print(f"\nTarget:\n{iris_dataset['target']}")
```
